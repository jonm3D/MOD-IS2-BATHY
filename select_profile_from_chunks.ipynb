```python
import geopandas as gpd
import os, sys, re
import pandas as pd

dir_chunks = "/Volumes/T7/Hawaii_Site_Assessments/ICESat2/Oahu/Oahu_20250606_141419/chunks"

parquet_files = [f for f in os.listdir(dir_chunks) if f.endswith('.parquet')]

# IS2_identifier 
```

```python
# read all files 
df_list = []
for file in parquet_files:
    df_temp = pd.read_parquet(os.path.join(dir_chunks, file), columns=['chunk_id', 'rgt', 'cycle', 'spot']).drop_duplicates()
    df_list.append(df_temp)

df = pd.concat(df_list, ignore_index=True)
```
